```python
import random


final_goal = "J"

max_iteration = 200


graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": ["G", "H", "I", "J"],
    "E": [],
    "F": [],
    "G": [],
    "H": [],
    "I": [],
    "J": []
}


successful_paths = []

total_epoch = 100
epoch = 0

success_state = 0

length_of_path = 0


while epoch < total_epoch:

    start_position = "A"

    iteration = 0

    path = [start_position]

    print("EPOCH:", epoch)

    while start_position != final_goal and iteration < max_iteration:

        print("EPOCH:----------------", epoch)

        print("Iteration", iteration)

        children = graph[start_position]

        if len(children) == 0:

            print("No more children")

            break

        random_direction = random.choice(children)

        start_position = random_direction

        print("MOVING TO =", start_position)

        path.append(start_position)

        iteration += 1

        print("TOTAL ITERATION", iteration)

    if start_position == final_goal:

        print("The walker has found the goal", path)

        print("The length of the path is:", len(path))

        length_of_path = length_of_path + len(path)

        successful_paths.append(path.copy())

        print("TOTAL LENGTH", length_of_path)

        success_state += 1

        print("SUCCESS STATE:", success_state)

        print("Probability:", success_state / (epoch + 1))

        print("Average steps:", length_of_path / success_state)

    else:

        print("Goal not found")

    epoch += 1

    print("TOTAL EPOCHS", epoch)
```
