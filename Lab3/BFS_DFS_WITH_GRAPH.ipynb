```python
graph = {
      "A": ["B", "C"],
      "B": ["D", "E"],
      "C": ["F"],
      "D": ["G", "H", "I", "J"],
      "E": [],
      "F": [],
      "G": [],
      "H": [],
      "I": [],
      "J": [],
  }
```

```python
def bfs_with_fifo(graph):



  print("Starting QUEUE",graph)

  queue = ["A"]
  visited = []

  while len(queue) != 0:

      node = queue[0]

      new_queue = []

      for i in range(1, len(queue)):
          new_queue.append(queue[i])

      queue = new_queue

      if node not in visited:

          visited.append(node)
          if node in graph:
              children = graph[node]
          else:
              children = []


          for child in children:

              if child not in visited and child not in queue:
                  queue.append(child)

      print("Removed:", node)
      print("\n")
      print("Updated Queue:", queue)

  print("\n")
  print("Final QUEUE STATUS:", queue)

  print("Final VISITED BFS:", visited)
```

```python
bfs_with_fifo(graph)
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}
print("Starting stack",graph)
print("\n\n")
stack = ["A"]
visited = []



while len(stack) != 0:


    node = stack[len(stack)-1]


    new_stack = []

    for i in range(0, len(stack)-1):
        new_stack.append(stack[i])

    stack = new_stack


    if node not in visited:

        visited.append(node)


        children = graph[node]

        for child in children:

            if child not in visited and child not in stack:
                stack.append(child)

    print("Visited:", visited)
    print("\n")
    print("Stack:", stack)

print("\n\n")
print("\nFinal DFS Traversal:", visited)
```
