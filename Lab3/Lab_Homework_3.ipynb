```python
queue = ["A", "B","C","D","E" ]
```

```python
def bfs_with_fifo(queue):
  print("Starting QUEUE",queue)

  while len(queue)!=0:

    node= queue[0]
    new_queue= []


    for i in range (1, len(queue)):
      new_queue.append(queue[i])



    queue = new_queue

    print("Removed:", node)
    print("Updated Queue:", queue)
```

```python
bfs_with_fifo(queue)
```

```python
stack = ['A', 'B', 'C', 'D', 'E']
```

```python
def dfs_with_lifo(stack):
  print("Starting QUEUE",stack)

  while len(stack) != 0:


      node = stack[len(stack)-1]


      new_stack = []



      for i in range(0, len(stack)-1):
        new_stack.append(stack[i])


      stack = new_stack

      print("Removed:", node)
      print("Updated Stack", stack)
```

```python
dfs_with_lifo(stack)
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}

queue = ["A"]
visited = []

while len(queue) != 0:

    node = queue[0]

    new_queue = []
    for i in range(1, len(queue)):
        new_queue.append(queue[i])

    queue = new_queue

    if node not in visited:
        visited.append(node)

        children = graph[node]

        for child in children:
            if child not in visited:
                queue.append(child)

    print("Visited:", node)
    print("Queue now:", queue)
    print("Updated Visited list:", visited)

```
