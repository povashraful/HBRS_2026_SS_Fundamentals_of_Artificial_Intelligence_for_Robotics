```python

def bfs_with_fifo(graph):



  print("Starting QUEUE",graph)

  queue = ["A"]
  visited = []

  while len(queue) != 0:

      node = queue[0]

      new_queue = []

      for i in range(1, len(queue)):
          new_queue.append(queue[i])

      queue = new_queue

      if node not in visited:

          visited.append(node)
          if node in graph:
              children = graph[node]
          else:
              children = []


          for child in children:

              if child not in visited and child not in queue:
                  queue.append(child)

      print("Removed:", node)
      print("\n")
      print("Updated Queue:", queue)

  print("\n")
  print("Final QUEUE STATUS:", queue)

  print("Final VISITED BFS:", visited)
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}
print("Starting stack",graph)
print("\n\n")
stack = ["A"]
visited = []



while len(stack) != 0:


    node = stack[len(stack)-1]


    new_stack = []

    for i in range(0, len(stack)-1):
        new_stack.append(stack[i])

    stack = new_stack


    if node not in visited:

        visited.append(node)


        children = graph[node]

        for child in children:

            if child not in visited and child not in stack:
                stack.append(child)

    print("Visited:", visited)
    print("\n")
    print("Stack:", stack)

print("\n\n")
print("\nFinal DFS Traversal:", visited)
```

```python
import random


final_goal = [5, 5]

max_iteration = 50


grid_size = 6
print(grid_size)

successful_paths = []
total_epoch = 10
epoch = 0
success_state = 0
length_of_path= 0

while epoch < total_epoch:

    start_position = [0, 0]
    iteration = 0
    path = [start_position.copy()]
    print("EPOCH:", epoch)


    while start_position != final_goal and iteration < max_iteration:

        user_input = random.choice(["up", "down", "left", "right"])
        print("EPOCH:----------------", epoch)
        print("Iteration", iteration)

        up = [start_position[0], start_position[1] + 1]
        down = [start_position[0], start_position[1] - 1]
        right = [start_position[0] + 1, start_position[1]]
        left = [start_position[0] - 1, start_position[1]]

        if user_input == "up":
            new_position = up
        elif user_input == "down":
            new_position = down
        elif user_input == "right":
            new_position = right
        else:
            new_position = left

        if 0 <= new_position[0] < grid_size and 0 <= new_position[1] < grid_size:
            start_position = new_position
            print("MOVING DIRECTION  =  ",user_input )
            print("NEW POSITION:", start_position)
        else:
            print("Out of boundary")

        path.append(start_position.copy())
        iteration += 1
        print("TOTAL ITERATION", iteration)

    if start_position == final_goal:
        print("The walker has found the goal", path)
        print("The length of the path is: ", len(path))
        length_of_path = length_of_path + len(path)
        successful_paths.append(path.copy())

        print("TOTAL LENGTH",length_of_path)
        success_state+=1
        print("SUCCESS STATE: ", success_state)
        print("Probability",epoch/success_state)
        print("Average steps it took to reach the goal",length_of_path/success_state)

    else:
        print("Goal not found")



    epoch += 1
    print("TOTAL EPOCHS", epoch)
```
